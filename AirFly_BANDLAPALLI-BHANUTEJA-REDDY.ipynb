```python
# ✅ Load the dataset
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load CSV
df = pd.read_csv("DelayedFlights.csv")

# 📤 Output: First few rows
df.head()
```

```python
# ✅ Dataset shape
df.shape
# 📤 Output: (num_rows, num_columns)

# ✅ Column info and data types
df.info()
# 📤 Output: Column names, non-null counts, data types

# ✅ Summary statistics
df.describe(include='all')
# 📤 Output: Count, mean, std, min, max, unique values for each column
```

```python
# ✅ Missing values
df.isnull().sum()
# 📤 Output: Count of missing values per column

# ✅ Duplicate rows
df.duplicated().sum()
# 📤 Output: Number of duplicate rows

# ✅ Data types
df.dtypes
# 📤 Output: Data type of each column
```

```python
# ✅ Histograms for numeric columns
df.hist(figsize=(15,10), bins=30)
plt.suptitle("Univariate Distribution of Numeric Features")
plt.tight_layout()
plt.show()
# 📤 Output: Histograms showing distribution of delays, airtime, distance, etc.

# ✅ Top categories for key categorical columns
for col in ['UniqueCarrier', 'Origin', 'Dest']:
    print(f"\nTop 5 values in {col}:\n", df[col].value_counts().head())
# 📤 Output: Frequency counts for top 5 categories
```

```python
# ✅ Correlation matrix
plt.figure(figsize=(12,8))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title("Correlation Matrix")
plt.show()
# 📤 Output: Heatmap showing correlation between numeric features

# ✅ Boxplot: Departure Delay by Carrier
plt.figure(figsize=(12,6))
sns.boxplot(x='UniqueCarrier', y='DepDelay', data=df)
plt.title("Departure Delay by Carrier")
plt.xticks(rotation=45)
plt.show()
# 📤 Output: Boxplot showing delay distribution per airline
```

```python
# ✅ IQR method for DepDelay
Q1 = df['DepDelay'].quantile(0.25)
Q3 = df['DepDelay'].quantile(0.75)
IQR = Q3 - Q1

outliers = df[(df['DepDelay'] < Q1 - 1.5 * IQR) | (df['DepDelay'] > Q3 + 1.5 * IQR)]
print("Outliers in DepDelay:", outliers.shape[0])
# 📤 Output: Number of outlier rows based on DepDelay
```

```python
# Count duplicates before dropping
before = df.duplicated().sum()
print(f"🔍 Duplicates before drop: {before}")

# Drop duplicates
df.drop_duplicates(inplace=True)

# Count duplicates after dropping
after = df.duplicated().sum()
print(f"✅ Duplicates after drop: {after}")
print(f"🧹 Rows removed: {before - after}")
```

```python
# ✅ Count missing values per column
missing_counts = df.isnull().sum()
missing_percent = (missing_counts / len(df)) * 100

# 📤 Output: Table of missing values
missing_summary = pd.DataFrame({
    'MissingCount': missing_counts,
    'MissingPercent': missing_percent.round(2)
})
missing_summary[missing_summary['MissingCount'] > 0].sort_values(by='MissingPercent', ascending=False)
```

```python
delay_cols = ['CarrierDelay', 'WeatherDelay', 'NASDelay', 'SecurityDelay', 'LateAircraftDelay']
df[delay_cols] = df[delay_cols].fillna(0)
```

```python
# Check if any missing values remain in delay columns
print("🔍 Remaining missing values in delay columns:")
print(df[delay_cols].isnull().sum())
```

```python
# ✅ Average contribution of each delay type
avg_delay_contrib = df[['CarrierDelay', 'WeatherDelay', 'NASDelay', 'SecurityDelay', 'LateAircraftDelay']].mean().sort_values(ascending=False)

# 📊 Bar plot
plt.figure(figsize=(10,6))
sns.barplot(x=avg_delay_contrib.index, y=avg_delay_contrib.values)
plt.title("Average Contribution of Delay Types")
plt.ylabel("Average Delay (minutes)")
plt.xticks(rotation=45)
plt.show()
```

```python
# ✅ Rename columns temporarily to match pandas expectations
df_rename = df.rename(columns={'Year': 'year', 'Month': 'month', 'DayofMonth': 'day'})

# ✅ Create FlightDate using renamed columns
df['FlightDate'] = pd.to_datetime(df_rename[['year', 'month', 'day']])

# 📤 Output: Confirm new datetime column
print(df[['Year', 'Month', 'DayofMonth', 'FlightDate']].head())
```

```python
# ✅ Confirm delay columns are numeric
delay_cols = ['DepDelay', 'ArrDelay', 'CarrierDelay', 'WeatherDelay', 'NASDelay', 'SecurityDelay', 'LateAircraftDelay']
non_numeric = [col for col in delay_cols if df[col].dtype not in [np.int64, np.float64]]

if non_numeric:
    print("⚠️ Non-numeric delay columns:", non_numeric)
else:
    print("✅ All delay columns are numeric and in minutes")
```

```python
# ✅ Extract weekday and route
df['Weekday'] = df['FlightDate'].dt.day_name()
df['Route'] = df['Origin'] + "-" + df['Dest']

# ✅ Add Season based on Month
def get_season(month):
    return ('Winter' if month in [12,1,2] else
            'Spring' if month in [3,4,5] else
            'Summer' if month in [6,7,8] else
            'Fall')
df['Season'] = df['Month'].apply(get_season)

# 📤 Output: Confirm new features
print(df[['FlightDate', 'Weekday', 'Route', 'Season']].head())
```

```python
# ✅ Remove negative departure delays
initial_rows = df.shape[0]
df = df[df['DepDelay'] >= 0]
final_rows = df.shape[0]

print(f"🧹 Removed {initial_rows - final_rows} rows with negative DepDelay")
```

```python
# ✅ Monthly average departure delay
monthly_delays = df.groupby(df['FlightDate'].dt.to_period('M'))['DepDelay'].mean().reset_index()
monthly_delays.columns = ['Month', 'AvgDepDelay']

# 📤 Output: Monthly delay trends
print(monthly_delays.head())
```

```python
# ✅ Save cleaned dataset
df.to_csv("Cleaned_DelayedFlights.csv", index=False)
print("✅ Cleaned dataset saved as 'Cleaned_DelayedFlights.csv'")
```
